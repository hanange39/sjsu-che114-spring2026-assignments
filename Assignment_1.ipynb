```python
# This shows the charging graphs of MacBook.

# import necessary libraries
import pandas as pd # Pandas is a library for data analysis. The name comes from 'panel data'. As the name "pandas" is long, it is replaced with "pd" in this line.
from pandas import read_csv # From pandas, import the function to read a csv file.
import matplotlib.pyplot as plt # Matplotlib is a plotting library. Pyplot is a collection of functions. In this line, it is shortened as "plt".

# read the CSV test data. assign it to "df". The first row in CSV, row 0, is automatically used as the header. In this code, manually specified header=0 for an education purpose.)
df = pd.read_csv('CHEMATE114_01_MacBook_charge_data.csv', header=0) #If read_csv is not imported earlier, then use 'pd.read_csv'. This runs read_csv under pd.
```

```python
# show what's inside 'df'.
df
```

```python
# plot current over time
plt.plot(df['Time (seconds)'],df['Charged capacity (Ah)'])
plt.xlabel('Time (seconds)')
plt.ylabel('Charged capacity (Ah)')
plt.show()
```

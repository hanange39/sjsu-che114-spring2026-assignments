```python
# This cell imports necessary dataset and libraries. Do not change it.
# Data source: Quinlan, R. (1993). Auto MPG [Dataset]. UCI Machine Learning Repository. https://doi.org/10.24432/C5859H.
!pip install ucimlrepo
from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# fetch dataset
auto_mpg = fetch_ucirepo(id=9)

# data (as pandas dataframes)
X = auto_mpg.data.features
y = auto_mpg.data.targets
```

```python
X #FYI. This displays X.
```

```python
y #FYI. This displays y.
```

```python
#Write the code to split the data into training and test sets.
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
#Write the code to select the RandomForestRegressor model.
model=RandomForestRegressor(n_estimators=1000, random_state=42)
```

```python
#Write the code to train the model with the training data.
model.fit(x_train, y_train)
```

```python
#Write the code to make predictions with the test data.
y_predict = model.predict(x_test)
```

```python
#Write the code to calculate MAE, using the test data and the predctions.
mean_absolute_error(y_test,y_predict)
```

```python
#FYI:Visualization of predicted and real y over one of the features: weight.
import matplotlib.pyplot as plt
plt.scatter(x_test['weight'], y_predict, label='Predicted mpg')
plt.scatter(x_test['weight'], y_test, label='Real mpg')
plt.legend()
plt.xlabel('weight')
plt.ylabel('mpg')
plt.title('Real vs Predicted mpg')
```

```python
# prompt: create a program that predicts Survived that is binary. use train.csv for training, use test.csv to predict Survived as binary. make the submission file that uses the same format as gender_submission.csv.

!pip install "flaml[automl]"
import pandas as pd
from sklearn.model_selection import train_test_split
from flaml import AutoML
from sklearn.metrics import accuracy_score

# Load the training data
train_df = pd.read_csv('train.csv')

# Fill missing age values with the median
train_df['Age'].fillna(train_df['Age'].median(), inplace=True)

# Convert 'Sex' to numerical values
train_df['Sex'] = train_df['Sex'].map({'female': 0, 'male': 1})

# Select features and target variable
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
target = 'Survived'

X_train = train_df[features]
y_train = train_df[target]

# Initialize and train with AutoML
model = AutoML(seed=42,
               time_budget=30,
               metric="accuracy",
               task="classification",
               n_splits=5
               )
model.fit(X_train, y_train)

# Load the test data
test_df = pd.read_csv('test.csv')

# Preprocess the test data similarly
test_df['Age'].fillna(test_df['Age'].median(), inplace=True)
test_df['Sex'] = test_df['Sex'].map({'female': 0, 'male': 1})
test_df['Fare'].fillna(test_df['Fare'].median(), inplace=True) # Fill missing fare values

# Make predictions
X_test = test_df[features]
predictions = model.predict(X_test)

# Create a submission DataFrame
submission_df = pd.DataFrame({'PassengerId': test_df['PassengerId'], 'Survived': predictions})

# Save the submission file
submission_df.to_csv('submission.csv', index=False)
```
